```python
# changing to 0 as no and 1 as yes 
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df["SeniorCitizen"] = df["SeniorCitizen"].apply(conv)  
```

```python
df.head(3) 
```

```python
plt.figure(figsize=(6,4))
ax = sns.countplot(data=df,x='Churn',color= 'green',alpha = 0.7,width=0.6)
ax.bar_label(ax.containers[0])
plt.title("Customer Churn Breakdown")
plt.show()
```

```python
ax = sns.countplot(data=df, x='gender', hue='Churn')

for container in ax.containers:
    ax.bar_label(container, padding=3) 

plt.title("Gender Churn Count Distribution",fontsize = 15,)
plt.show()
```

```python
gb = df.groupby("Churn").agg({"Churn":"count"})
gb

total = gb["Churn"].sum()

plt.pie(
    gb["Churn"],
    labels=gb.index,
    autopct=lambda p: f'{p:.1f}%\n({int(p*total/100)})',
    wedgeprops={"edgecolor": "black"}
) 


plt.title(
    'Customer Churn Distribution',
    fontsize=16,
    fontweight='bold',
    pad=20
)

plt.legend()
plt.tight_layout() 
plt.show() 
```

#from the given pie chart we can conclude that 26.54% of our customers as churned out.
# now lets explore the reason behind it. 

```python
plt.figure(figsize=(5,4))
ax = sns.countplot(data=df, x='SeniorCitizen', hue='Churn')

ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.title("Senior Citizen churn count distribution")
plt.show()  
```

```python
ct = pd.crosstab(df['SeniorCitizen'], df['Churn'])

ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100

ax = ct_pct.plot(
    kind='bar',
    stacked=True, 
    figsize=(5,4),
    color=['skyblue', 'orange']
)

for container in ax.containers:
    labels = [f'{v:.1f}%' if v > 0 else '' for v in container.datavalues]
    ax.bar_label(container, labels=labels, label_type='center')

plt.title('Senior Citizen Churn Distribution (%)')
plt.ylabel('Percentage')
plt.show()
```

Among non-senior citizens, 24% have churned and 76% have stayed.

Among senior citizens, 42% have churned and 58% have stayed.

 Since the churn percentage for senior citizens (42%) is much higher than for non-senior citizens (24%), senior citizens are more likely to leave the service. 

```python
sns.histplot(x="tenure",data=df,bins=72, hue="Churn")
plt.title("Tenure Churn Analysis")
plt.show() 
```

The probability of churn appears to decrease with customer loyalty and longer relationships.

Customers with very high tenure show strong brand loyalty and low churn risk.

Retention programs should primarily target customers during their early lifecycle rather than long-term customers.

"Tenure exhibits a strong relationship with churn. Customers with lower tenure show substantially higher churn rates, while customers with longer tenure are more likely to be retained. The highest concentration of churn occurs among new customers, suggesting that customer onboarding and early engagement strategies are crucial for improving retention."


```python
plt.figure(figsize=(4,5))
sns.boxplot(data=df, x='Churn', y='tenure', hue='Churn', width=0.5,palette=['#4C72B0', '#DD8452'])
plt.title('Tenure vs Churn')
plt.show()
```

Tenure shows a strong negative relationship with churn. Customers who churn have a significantly lower median tenure (approximately 10 months) compared to retained customers (approximately 38 months). This suggests that customers are most vulnerable to churn during the early stages of their relationship with the company. As tenure increases, the likelihood of churn decreases, indicating stronger customer loyalty among long-term subscribers. 



The median tenure for Churn = Yes is around 10 months.
The median tenure for Churn = No is around 38 months.

This indicates that customers who leave tend to do so relatively early.

```python
plt.figure(figsize=(7,5))

ax = sns.countplot(
    data=df,
    x='Contract',
    hue='Churn',
    palette='Set2'
)

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.title(
    'Customer Churn by Contract Type',
    fontsize=16,
    fontweight='bold',
    color='black',
    fontstyle='italic',
    loc='center',
    pad=20,
    backgroundcolor='lightyellow'
) 
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
columns = ['PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 
           'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

# Number of columns for the subplot grid (you can change this)
n_cols = 3
n_rows = (len(columns) + n_cols - 1) // n_cols  # Calculate number of rows needed

# Create subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 4))  # Adjust figsize as needed

# Flatten the axes array for easy iteration (handles both 1D and 2D arrays)
axes = axes.flatten() 

# Iterate over columns and plot count plots
for i, col in enumerate(columns):
    sns.countplot(x=col, data=df, ax=axes[i], hue = df["Churn"])
    axes[i].set_title(f'Count Plot of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')


for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()  
```
